```python
"""
Write a Python function that accepts two integer numbers. If the product of the two numbers is less than or equal
to 1000, return their product; otherwise, return their sum.
"""
def p_sum(a,b):
    result = a * b

    if result <= 1000:
        return a * b
    else:
        return a + b
```

```python
p_sum(100,100)
```

```python
"""
Iterate through the first 10 numbers (0–9). In each iteration, print the current number, the previous number, and
their sum.
"""

print("Printing current and previous number sum in a range(10)")

previous_n = 0

for i in range(10):
    s = previous_n + i
    print(f"Current Number {i} Previous Number {previous_n} Sum: {s}" )

    previous_n = i

```

```python
"""
Write a function to return True if the first and last number of a given list is the same. If the numbers are different,
return False.

Given Input:

numbers_x = [10, 20, 30, 40, 10]
numbers_y = [75, 65, 35, 75, 30]

Expected Output:

Given list: [75, 65, 35, 75, 30] | result is False
Given list: [10, 20, 30, 40, 10] | result is True
"""

def list_same(list):

    if list[0] == list[-1]:
        return True
    else:
        return False
```

```python
"""
Write a program to extract each digit from an integer in the reverse order.
"""
def reverse_number(num):
    
    str_num = str(num)
    reverse_num = str_num[::-1]
    num = int(reverse_num)
    return num
```

```python
"""
Calculate income tax for a given income based on these rules:

-> First $10,000: 0% tax
-> Next $10,000: 10% tax
-> Remaining income: 20% tax
"""

def income_tax(income):
    if income <= 10000:
        tax = 0

    elif income <= 20000:
        tax = (income - 10000) * 0.10

    else:
        tax = 10000 * 0.10 + (income - 20000) * 0.20

    print(f"Total income tax to pay is {tax}")
    
```

```python
income_tax(45000)
```

```python
"""
Start with a list of 10 numbers. Iterate through them and sort them into two separate lists: 
one for even numbers and one for odd numbers.

Given Input: numbers = [12, 7, 34, 21, 5, 10, 8, 3, 19, 2]

Expected Output:

Even numbers: [12, 34, 10, 8, 2]
Odd numbers: [7, 21, 5, 3, 19]
"""

numbers = [12, 7, 34, 21, 5, 10, 8, 3, 19, 2]

even = []
odd = []

for i in numbers:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print(f"Even numbers: {even}")
print(f"Odd numbers: {odd}")
```

```python
"""
This exercise introduces “Metadata Extraction.” Often, you aren’t just interested in the 
data itself, but in its properties. In web development, this logic is used to validate 
if a user’s password or username meets specific length requirements.

Given Input: words = ["Apple", "Banana", "Cherry", "Date", "Elderberry"]

Expected Output:

Apple - 5 Banana - 6 Cherry - 6 Date - 4 Elderberry - 10
"""

words = ["Apple", "Banana", "Cherry", "Date", "Elderberry"]

for word in words:
    length = len(word)
    print(f"{word} - {length}", end = " ")
```

```python
"""
Write a program that counts how many times each word appears in a given paragraph and 
stores these counts in a dictionary.

Given Input: text = "apple banana apple cherry banana apple"

Expected Output: {'apple': 3, 'banana': 2, 'cherry': 1}
"""
text = "apple banana apple cherry banana apple"

counts = {}

words = text.split()

for word in words:
    if word in counts:
        counts[word]+=1
    else:
        counts[word] = 1

print(counts)
```

```python
"""
Write a program to check if a user-entered string contains any numeric digits. Use a for loop
to examine each character.

Given Input: input_string = "Python3"

Expected Output: The string 'Python3' contains digits: True

"""

input_string = input("Enter your input:")
contains_digit = False

for d in input_string:
    if d.isdigit():
        contains_digit = True
        break
print(f"The string '{input_string}' contains digits: {contains_digit}")
```

```python
"""
Create a countdown timer that starts from a given number and counts down to zero using
a while loop

Given Input: start_count = 5

Expected Output: 5 4 3 2 1 Blast off!
"""

start_count = int(input("Enter  the number to  begin with:"))

while start_count <= start_count:
    print(start_count,end = " ")

    if start_count == 1:
        break
    start_count-=1
print(end = "Blast off")
```

```python
"""
Create a Car class with attributes for make, model, and year. Include a method called 
start_engine() that prints a formatted string describing the car starting up.

Given Input: Make: “Toyota”, Model: “Camry”, Year: 2022

Expected Output: The 2022 Toyota Camry's engine is now running!
"""

class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def car_engine(self):
        print(f"The {self.year} {self.make} {self.model} engine is now running!")

my_car = Car("Toyota", "Camry's", 2022)
my_car.car_engine()
```
