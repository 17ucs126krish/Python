# Lambda Function Inside Another Function

```python
def power(x):
    return lambda n : x ** n
```

```python
cube = power(2)(4)
```

```python
print(cube)
```

```python
"""
#4) Flatten the following list to a new list:

countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output:
[['FINLAND','FIN', 'HELSINKI'], ['SWEDEN', 'SWE', 'STOCKHOLM'], ['NORWAY', 'NOR', 'OSLO']]

"""
```

```python
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
```

```python
c_codes_capital = [[country.upper(), country[:3].upper(), capital.upper()] for country in countries for (country, capital) in country ]

c_codes_capital
```

```python
"""
#5) Change the following list to a list of dictionaries:

countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]

output:
[{'country': 'FINLAND', 'city': 'HELSINKI'},
{'country': 'SWEDEN', 'city': 'STOCKHOLM'},
{'country': 'NORWAY', 'city': 'OSLO'}]
"""
```

```python
country_city = [{'country' : country.upper(), 'city' : city.upper()} for country in countries for (country , city) in country]
country_city
```

```python
"""
#6) Change the following list of lists to a list of concatenated strings:

names = [[('Asabeneh', 'Yetayeh')], [('David', 'Smith')], [('Donald', 'Trump')], [('Bill', 'Gates')]]

output
['Asabeneh Yetaeyeh', 'David Smith', 'Donald Trump', 'Bill Gates']
"""
```

```python
names = [[('Asabeneh', 'Yetayeh')], [('David', 'Smith')], [('Donald', 'Trump')], [('Bill', 'Gates')]]
```

```python
full_name = [[" ".join((first_name, last_name))] for name in names for (first_name, last_name) in name]
full_name
```
