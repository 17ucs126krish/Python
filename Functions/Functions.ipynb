```python
"""
#7) Quadratic equation is calculated as follows: ax² + bx + c = 0. Write a function which calculates solution set of a quadratic equation, solve_quadratic_eqn.
"""

import math

def solve_quadratic_eqn(a,b,c):
    # Step 1: Checking whether it is a quadratic equation
    
    if a == 0:
        return "Not a quadratic equation (a cannot be 0)"
    #Step 2: Calculating discriminant
    
    D = b ** 2 -  4 * a *c
    
    #Step 3: Solving using discriminant
    
    if D > 0:
        x1 = (-b + math.sqrt(D)) / (2 * a)
        x2 = (-b - math.sqrt(D)) / (2 * a)
        return x1 , x2
    elif D == 0:
        x = - b/(2*a)
        return x
    else:
        return "No real roots"
        
```

```python
solve_quadratic_eqn(1,-3, 2)
```

```python
solve_quadratic_eqn(1,2,1)
```

```python
solve_quadratic_eqn(1,1,1)
```

```python
solve_quadratic_eqn(0, 2, 1)
```

```python
"""
#4) Write different functions which take lists. They should calculate_mean, calculate_median, 
calculate_mode, calculate_range, calculate_variance, calculate_std (standard deviation).
"""

num = [1,2,3,4,5]

def calculate_mean(n):
    return sum(num)/len(num)
```

```python
calculate_mean(num)
```

```python
def calculate_median(n):
    sorted_nums = sorted(n)
    n = len(sorted_nums)
    mid = n//2
    
    if n % 2 ==1:
        return sorted_nums[mid]
    else:
        return (sorted_nums[mid-1] + sorted_nums[mid])/2
        
```

```python
calculate_median(num)
```

```python
def calculate_mode(n):
    frq = {}
    for num in n:
        frq[num] = frq.get(num, 0) + 1
    
    max_count = max(frq.values())
    
    m = []

    for k, v in frq.items():
        if v == max_count:
            m.append(k)
    return m
            
        
    
```

```python
a = [1,2,3,4,4,5,6,7,8,8,8,9]
```

```python
calculate_mode(a)
```

```python
def calculate_range(n):
    return max(n) - min(n)
```

```python
def calculate_variance(n):
    mean = calculate_mean(num)
    total = 0
    
    for i in n:
        total = total + (i - mean) ** 2
    
    return total /len(n)
```

```python
calculate_variance(num)
```

```python
import math

def calculate_std(n):
    v = calculate_variance(n)
    return math.sqrt(v)
    
```

```python
"""
#3) Write a function which checks if all the items of the 
list are of the same data type.
"""
def same_type(lst):
    #If its empty return False
    if not lst:
        return False
    
    first_type = type(lst[0])
    
    #Checks wether the each element is of same type by comparing the data type of the first element
    for item in lst:
        if type(item) != first_type:
            return False
    
    return True
    
```

```python
same_type([1,2,3,4,5])
```

```python
"""
#4) Write a function which check if provided variable is a valid python variable
"""

import keyword

def is_variable(name):
    if not isinstance(name, str):
        return False
    
    if not name:
        return False
    
    if not name.isidentifier():
        return False
    
    if keyword.iskeyword(name):
        return False
    
    return True
```

```python
"""
(i) Create a function called the most_spoken_languages in the world. It should return 10 or 20 most spoken languages 
in the world in descending order.
"""
```

```python
def most_spoken_languages(data):
    
    popular_language = {}
    
    for country in data:
        for lang in country['languages']:
            if lang in popular_language:
                
                popular_language[lang]+=1
            else:
                popular_language[lang] = 1
    
    import pandas as pd
    
    df = pd.DataFrame(popular_language.items(), columns = ['Languages', 'Counts'])
    
    top_10_languages = df.sort_values(by = "Counts", ascending = False).head(10)
    
    return top_10_languages
```

```python
"""
(ii) Create a function called the most_populated_countries. It should return 10 or 20 most populated countries in
descending order. 
"""
```

```python
def most_populated_countries(data):
    
    p_countries = {}
    for country in data:
        country_name = country['name']
        population = country['population']
        p_countries[country_name] = population
    
    import pandas as pd
    
    df = pd.DataFrame(p_countries.items(), columns = ['country_name', 'population'])
    
    top_10_p_countries = df.sort_values(by = 'population', ascending = False).head(10)
    
    return top_10_p_countries
        
```
